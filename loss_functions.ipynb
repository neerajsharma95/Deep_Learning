**Mean Absolute Error (MAE)**

```python
import pandas as pd
import tensorflow as tf

# Read Dataset
data = pd.read_csv("student_file.csv")
print(data.head())

X = data[['StudyHours']].values
y = data['Marks'].values

# Forward Pass Model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(
        units=1,
        activation='sigmoid',
        input_shape=(1,)
    )
])

# Forward Pass
prediction = model(X)
print("Predictions")
print(prediction.numpy())

# Calculate MAE
mae = tf.reduce_mean(tf.abs(y.reshape(-1,1) - prediction))

print("\nMAE Loss")
print(mae.numpy())
```

**Mean Squared Error**

```python
import pandas as pd
import tensorflow as tf

# Read Dataset
data = pd.read_csv("student_file.csv")
print(data.head())

X = data[['StudyHours']].values
y = data['Marks'].values

# Forward Pass Model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(
        units=1,
        activation='sigmoid',
        input_shape=(1,)
    )
])

# Forward Pass
prediction = model(X)

print("Predictions")
print(prediction.numpy())

# Calculate MSE
mse = tf.reduce_mean(
    tf.square(y.reshape(-1,1)-prediction)
)

print("MSE =",mse.numpy())
```

**Huber's Loss**

```python
import pandas as pd
import tensorflow as tf

data = pd.read_csv("student_file.csv")

X = data[['StudyHours']].values
y = data['Marks'].values

model = tf.keras.Sequential([
    tf.keras.layers.Dense(
        units=1,
        activation='sigmoid',
        input_shape=(1,)
    )
])

prediction = model(X)

huber = tf.keras.losses.Huber(delta=1.0)

loss = huber(y.reshape(-1,1),prediction)

print("Huber Loss")
print(loss.numpy())
```

**Binary Cross Entropy**

```python
import pandas as pd
import tensorflow as tf

data = pd.read_csv("student_result.csv")
print(data.head())

X = data[['Study','Sleep','Play']].values
y = data['Result'].values

model = tf.keras.Sequential([

    tf.keras.layers.Dense(
        units=1,
        activation='sigmoid',
        input_shape=(3,)
    )

])

prediction = model(X)

print("Predictions")
print(prediction.numpy())

bce = tf.keras.losses.BinaryCrossentropy()

loss = bce(y.reshape(-1,1),prediction)

print("\nBinary Cross Entropy")
print(loss.numpy())
```

**Categorical Cross Entropy**

```python
import pandas as pd
import tensorflow as tf

data = pd.read_csv("iris.csv")

X = data[['Length',
          'Width',
          'PetalLength',
          'PetalWidth']].values

y = tf.keras.utils.to_categorical(
        data['Class'],
        num_classes=3
)

model = tf.keras.Sequential([

    tf.keras.layers.Dense(
        units=3,
        activation='softmax',
        input_shape=(4,)
    )

])

prediction = model(X)

print(prediction.numpy())

cce = tf.keras.losses.CategoricalCrossentropy()

loss = cce(y,prediction)

print("\nCategorical Cross Entropy")

print(loss.numpy())
```
