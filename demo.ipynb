```python
import pandas as pd
# Read Dataset
data = pd.read_csv("student_data.csv")
data.head()
```

```python
X = data[['StudyHours',
          'SleepHours',
          'PlayingHours']]
y = data['Pass']
print(X)
print(y)
```

```python
# Single neuron and step activation function
import tensorflow as tf

model = tf.keras.Sequential([

    tf.keras.layers.Dense(
        units=1,
        activation='linear',
        input_shape=(3,)
    )

])

prediction = model(X)

step_output = tf.where(prediction > 0, 1, 0)

print(step_output)
```

```python
# Single Neuron, Sigmoid activation 
import tensorflow as tf

model = tf.keras.Sequential([

    tf.keras.layers.Dense(
        units=1,
        activation='sigmoid',
        input_shape=(3,)
    )

])

output = model(X)

print(output)
```

```python
#multi neuron step activation
import tensorflow as tf

model = tf.keras.Sequential([

    tf.keras.layers.Dense(
        units=4,
        activation='linear',
        input_shape=(3,)
    ),

    tf.keras.layers.Dense(
        units=1,
        activation='linear'
    )

])

prediction = model(X)

step_output = tf.where(prediction > 0, 1, 0)

print(step_output)
```

```python
# Multi neuron, sigmoid activation
import tensorflow as tf

model = tf.keras.Sequential([

    tf.keras.layers.Dense(
        units=4,
        activation='sigmoid',
        input_shape=(3,)
    ),

    tf.keras.layers.Dense(
        units=1,
        activation='sigmoid'
    )

])

output = model(X)

print(output)
```
